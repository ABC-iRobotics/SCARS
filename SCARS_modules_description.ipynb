# <center>Description of the modules</center>

| Modul names                   | Description                                             |
|-------------------------------|---------------------------------------------------------|
| data_manipulation_module      | Functions for data manipulation                         |
| nts_classification_module     | Functions for non-time series classification algorithms |
| ts_classification_module      | Functions for time series classification algorithms     |
| validation_and_results_module | Functions for validation and results                    |
| param_tuning_module           | Functions for parameter tuning                          |
| dim_reduction_module          | Functions for dimensionality reduction                  |
| plots_module                  | Functions for plotting                                  |

<br></br>
### <center>data_manipulation_module</center>

| Function                                  | Description                                                                               | Parameter list                                                                                 |
|-------------------------------------------|-------------------------------------------------------------------------------------------|------------------------------------------------------------------------------------------------|
| Loading_data                              | Loading the data                                                                          | Loading_data(files, separator, col_names)                                                      |
| Multi_class_target_variable               | Transforming the format of the target variable for multi-class classification             | Multi_class_target_variable(df_class_value, class_tresholds)                                   |
| Multi_target_variable                     | Creating classes for multiple target variables (based on mean)                            | Multi_target_variable(df, col_idx)                                                             |
| Approximate_entropy                       | Transforming time series data with ApEn                                                   | Approximate_entropy(df_list, return_ApEn)                                                      |
| Target_append                             | Attaching the target variable to the kinematic data after the ApEn transformation         | Target_append(df_class, return_ApEn, class_column_name)                                        |
| Features_target                           | Decomposition of kinematic data into features and target variable                         | Features_target(df)                                                                            |
| Train_test                                | Splitting the dataset into training and test sets based on a given test set ratio         | Train_test(X, Y, t_size)                                                                       |
| Standardization                           | Standardization of non-time series data                                                   | Standardization(X_train, X_test)                                                               |
| TS_df_append_to_max_len                   | Converting time series data to a uniform length (maximum length)                          | TS_df_append_to_max_len(df_list, feature_names)                                                |
| TS_df_append_truncate_to_avg_len          | Converting time series data to a uniform length (average length)                          | TS_df_append_truncate_to_avg_len(df_list, feature_names)                                       |
| TS_df_truncate_to_min_len                 | Converting time series data to a uniform length (minimum length)                          | TS_df_truncate_to_min_len(df_list)                                                             |
| TS_Y                                      | Defining the target variable for time series classifiers                                  | TS_Y(df_class, column_index)                                                                   |
| TS_Standardization                        | Standardization of time series data                                                       | TS_Standardization(X_train, X_test, feature_names)                                             |
| TS_kNN_DTW_transform                      | Transforming the format of time series data for the time series k-NN classifier           | TS_kNN_DTW_transform(X_train, X_test)                                                          |
| TS_NN_transform                           | Transforming the format of time series data for the NN classifier                         | TS_NN_transform(X_train, X_test)                                                               |
| Gestures_data_splitting                   | Splitting the dataset into gestures                                                       | Gestures_data_splitting(number_of_gestures, feature_names, gestures, gesture_df_list, df_list) |
| Gestures_ApEn_transform_target_append     | Transforming the gesture dataset with ApEn                                                | Gestures_ApEn_transform_target_append(number_of_gestures, df_gestures, feature_names, df_classes, classes_target_idx)             |
| Gestures_features_target                  | Decomposition of kinematic, gesture data into features and target variable                | Gestures_features_target(task_gestures, gestures_idxs, minimum_rows)                           |
| Gestures_train_test                       | Splitting the gesture dataset into training and test sets based on a given test set ratio | Gestures_train_test(X, Y, t_size)                                                              |
| Gestures_standardization                  | Standardization of the gesture dataset (non-time series)                                  | Gestures_standardization(gestures_X_train, gesture_X_test)                                     |
| TS_gestures_df_append_to_max_len          | Converting time series, gesture data to a uniform length (maximum length)                 | TS_gestures_df_append_to_max_len(df_list, feature_names)                                       |
| TS_gestures_df_append_truncate_to_avg_len | Converting time series, gesture data to a uniform length (average length)                 | TS_gestures_df_append_truncate_to_avg_len(df_list, feature_names)                              |
| TS_gestures_df_truncate_to_min_len        | Converting time series, gesture data to a uniform length (minimum length)                 | TS_gestures_df_truncate_to_min_len(df_list)                                                    |
| TS_gestures_X_Y_transform                 | Removing empty gestures, setting the format of the gestures                               | TS_gestures_X_Y_transform(gestures_df, df_class, column_index)                                 |
| TS_gestures_standardization               | Standardization of the gesture dataset (time series)                                      | TS_gestures_standardization(X_train, X_test, feature_names)                                    |
| TS_gesture_kNN_DTW_transform              | Transforming the time-series gestures data format for the time-series k-NN classifier     | TS_gesture_kNN_DTW_transform(X_train, X_test)                                                  |
| TS_gesture_NN_transform                   | Transforming the time-series gestures data format for the time-series NN classifier       | TS_gesture_NN_transform(X_train, X_test)                                                       |
| Best_hyperparameter_tuning_scores         | Displaying the best results of hyperparameter tuning.                                     | Best_hyperparameter_tuning_scores(score, estimator, cvs, test_size, class_alg)                 |
| Best_dimRed_hyperparameter_tuning_scores  | Displaying the best results of hyperparameter tuning with MI dimension reduction          | Best_dimRed_hyperparameter_tuning_scores(score, estimator, feature, threshold, cvs, test_size, class_alg)                                  |
